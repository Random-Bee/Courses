```python
import numpy as np
import matplotlib.pyplot as plt

# Define function to save figure as JPEG
def screen2jpeg(filename):
    plt.savefig(filename, dpi=300, format='jpeg', bbox_inches='tight')
    plt.draw()
```

```python
def plot_contour(tid, type=None, num_t=None):
    # Read data
    file_name = 'T_x_y_'
    if type != None:
        file_name = file_name + type + '_'
    if num_t != None:
        file_name = file_name + f'{num_t:02d}_'
    file_name = file_name + f'{tid:06d}.dat'
    data = np.loadtxt(file_name)

    # Reshape data
    n = int(np.sqrt(data.shape[0]))
    x = data[::n, 0]
    y = data[:n, 1]
    T = data[:, 2].reshape((n, n))

    # Contour plot
    plt.figure()
    plt.contourf(x, y, T.T, cmap='jet')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f't = {tid:06d}')
    plt.xlim([-0.05, 1.05])
    plt.ylim([-0.05, 1.05])
    plt.clim([-0.05, 1.05])
    plt.colorbar()
    plt.gca().tick_params(labelsize=14)
    file_name = 'cont_'
    if type != None:
        file_name = file_name + type + '_'
    file_name = file_name + f'T_{tid:06d}'
    if num_t != None:
        file_name = file_name + f'_{num_t:02d}'
    file_name = file_name + '.png'
    screen2jpeg(file_name)
    plt.close()
```

```python
def plot_combined_contours(tid, type):
    fig, axes = plt.subplots(2, 2, figsize=(10, 10))
    file_patterns = [
        f'T_x_y_{tid:06d}.dat',
        f'T_x_y_{type}_{tid:06d}.dat',
        f'T_x_y_{type}_{tid:06d}_02.dat',
        f'T_x_y_{type}_{tid:06d}_04.dat',
    ]

    titles = [
        f'Gauss_Seidel_{tid:06d}',
        f'{type.upper()}_{tid:06d}',
        f'{type.upper()}_{tid:06d}_02',
        f'{type.upper()}_{tid:06d}_04',
    ]

    for ax, file_name, title in zip(axes.flat, file_patterns, titles):
        if file_name and title:
            try:
                data = np.loadtxt(file_name)
                n = int(np.sqrt(data.shape[0]))
                x = data[::n, 0]
                y = data[:n, 1]
                T = data[:, 2].reshape((n, n))

                contour = ax.contourf(x, y, T.T, cmap='jet')
                ax.set_title(title, fontsize=12)
                ax.set_xlim([-0.05, 1.05])
                ax.set_ylim([-0.05, 1.05])
                ax.tick_params(labelsize=10)
                fig.colorbar(contour, ax=ax, shrink=0.75)
            except Exception as e:
                ax.set_title(f"Missing: {title}", fontsize=12)
                print(f"Error loading {file_name}: {e}")

    plt.tight_layout()
    screen2jpeg(f'cont_combined_{tid:06d}_{type}.png')
    plt.close()
```

```python
def plot_line(tid, type=None, num_t=None):
    # Read data
    file_name = 'T_x_y_'
    if type != None:
        file_name = file_name + type + '_'
    if num_t != None:
        file_name = file_name + f'{num_t:02d}_'
    file_name = file_name + f'{tid:06d}.dat'
    data = np.loadtxt(file_name)

    # Reshape data
    n = int(np.sqrt(data.shape[0]))
    x = data[::n, 0]
    y = data[:n, 1]
    T = data[:, 2].reshape((n, n))

    # Line plot at mid-y
    plt.figure()
    Tmid = T[:, n // 2]
    plt.plot(x, Tmid, '-', linewidth=2)
    plt.xlabel('x')
    plt.ylabel('T')
    plt.title(f'Profile along mid-y at t={tid:06d}')
    plt.xlim([-0.05, 1.05])
    plt.gca().tick_params(labelsize=14)
    file_name = 'line_midy_'
    if type != None:
        file_name = file_name + type + '_'
    file_name = file_name + f'T_{tid:06d}'
    if num_t != None:
        file_name = file_name + f'_{num_t:02d}'
    file_name = file_name + '.png'
    screen2jpeg(file_name)
    plt.close()
```

```python
def plot_combined_lines(tid, type=None):
    plt.figure(figsize=(8, 6))
    
    file_patterns = [
        (f'T_x_y_{tid:06d}.dat', 'Gauss Seidel'),
        (f'T_x_y_{type}_{tid:06d}.dat', f'{type.upper()}'),
        (f'T_x_y_{type}_{tid:06d}_02.dat', f'{type.upper()}_02'),
        (f'T_x_y_{type}_{tid:06d}_04.dat', f'{type.upper()}_04')
    ]

    for file_name, label in file_patterns:
        data = np.loadtxt(file_name)
        n = int(np.sqrt(data.shape[0]))
        x = data[::n, 0]
        y = data[:n, 1]
        T = data[:, 2].reshape((n, n))

        # Extract mid-y profile
        Tmid = T[:, n // 2]
        plt.plot(x, Tmid, label=label, linewidth=2)

    plt.xlabel('x')
    plt.ylabel('T')
    plt.title(f'Profile along mid-y at t={tid:06d}')
    plt.xlim([-0.05, 1.05])
    plt.legend()
    plt.gca().tick_params(labelsize=14)
    
    file_name = f'line_midy_combined_{tid:06d}_{type}.png'
    plt.savefig(file_name, dpi=300)
    plt.close()
```
