```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def screen2jpeg(filename):
    """Save the current figure as a JPEG file with screen dimensions."""
    plt.savefig(filename, format='jpeg', dpi=300, bbox_inches='tight')
    plt.draw()
```

```python
def plot_contour(tid, p = None):
    """Load data from file and generate a contour plot."""
    if p == None:
        filename = f'T_x_y_{tid:06d}.dat'
    else:
        filename = f'T_x_y_{tid:06d}_par_{p:02d}.dat'
    data = np.loadtxt(filename)
    
    n = int(np.sqrt(data.shape[0]))
    x = data[::n, 0]
    y = data[:n, 1]
    T = data[:, 2].reshape(n, n)
    
    plt.figure(figsize=(6, 5))
    contour = plt.contourf(x, y, T.T, cmap='jet')
    plt.colorbar(contour)
    
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f't = {tid:06d}')
    plt.xlim([-0.05, 1.05])
    plt.ylim([-0.05, 1.05])
    plt.clim(-0.05, 1.05)
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    
    if p == None:
        output_filename = f'cont_T_{tid:06d}.jpeg'
    else:
        output_filename = f'cont_T_{tid:06d}_{p:02d}.jpeg'
    screen2jpeg(output_filename)
    plt.show()
```

```python
plot_contour(0)
```
