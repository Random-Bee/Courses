```python
import numpy as np
import matplotlib.pyplot as plt

xg = np.loadtxt("xgrid.dat")
yg = np.loadtxt("ygrid.dat")
ff = np.loadtxt("flags.dat")

X, Y = np.meshgrid(xg, yg)

plt.figure(figsize=(8, 6))
plt.contourf(X, Y, ff.T, cmap="coolwarm")
plt.xlabel("x")
plt.ylabel("y")
plt.colorbar(label="Inside (1) / Outside (0)")
plt.title("Monte Carlo Circle Approximation")
plt.savefig(f"flags_{len(xg):04d}.png", dpi=300)
plt.show()
```

```python
error_data = np.loadtxt("error_n.dat")
n_values = error_data[:, 0]
errors = error_data[:, 1]
plt.figure(figsize=(8, 6))
plt.loglog(n_values, errors, "o", markersize=6, label="Observed Error")
plt.loglog(n_values, n_values**-1, "k--", label="O(1/n)")
plt.loglog(n_values, n_values**-2, "r--", label="O(1/n²)")
plt.xlabel("n")
plt.ylabel("Error")
plt.legend()
plt.title("Error vs. Grid Resolution (n)")
plt.savefig("error_n.png", dpi=300)
plt.show()
```
